# Part E: Data **Preprocessing**

```python
# Fix data anomaly where movie duration was placed into the rating column
misplaced_mask = df['duration'].isnull() & df['rating'].str.contains('min', na=False)
df.loc[misplaced_mask, 'duration'] = df.loc[misplaced_mask, 'rating']
df.loc[misplaced_mask, 'rating'] = np.nan

# Handle missing values
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna(df['country'].mode()[0])
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
df['date_added'] = df['date_added'].fillna(df['date_added'].mode()[0])

# Verify missing values
df.isnull().sum()
```

* director & cast: Contain missing values (2,634 and 825 missing respectively). Imputing with mode would create artificial dominance of a single creator. Replacing missing entries with 'Unknown' preserves data integrity without making false assumptions.

* country & rating: Imputed using the Mode (most frequent category), as these represent categorical metadata.

* duration / rating Anomaly: 3 rows (Louis C.K. titles) have duration data misplaced in the rating column. We first correct this data placement before handling missing values

```python
import numpy as np

# Extract numeric duration (minutes) for movies
df_movies = df[df['type'] == 'Movie'].copy()

df_movies['duration_min'] = df_movies['duration'].str.extract(r'(\d+)').astype(float)

# Remove missing duration values before calculating IQR
df_movies = df_movies.dropna(subset=['duration_min'])

# Calculate IQR bounds
Q1 = df_movies['duration_min'].quantile(0.25)
Q3 = df_movies['duration_min'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Cap outliers (Winsorization)
df_movies['duration_capped'] = np.clip(
    df_movies['duration_min'],
    lower_bound,
    upper_bound
)

# Display results
print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")

print(
    f"Original Max: {df_movies['duration_min'].max()} "
    f"-> Capped Max: {df_movies['duration_capped'].max()}"
)

print(
    f"Original Min: {df_movies['duration_min'].min()} "
    f"-> Capped Min: {df_movies['duration_capped'].min()}"
)
```

* Feature Selection: We extract numeric runtimes (in minutes) from the duration column for Movies.

* IQR Method: Runtimes shorter than 46.5 minutes or longer than 154.5 minutes are identified as statistical outliers.

```python
from sklearn.preprocessing import LabelEncoder

# Label Encoding for 'type'
le_type = LabelEncoder()
df['type_encoded'] = le_type.fit_transform(df['type'])

# Label Encoding for 'rating'
le_rating = LabelEncoder()
df['rating_encoded'] = le_rating.fit_transform(df['rating'])

df[['type', 'type_encoded', 'rating', 'rating_encoded']].head()
```

* type (Binary Variable): Encoded using Label Encoding (Movie $\rightarrow$ 0, TV Show $\rightarrow$ 1), since it is a binary classification feature.
* Rating (Nominal Variable): Encoded into numeric integer representation via Label Encoding or One-Hot Encoding to allow machine learning models to process categorical labels.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Extract numeric duration for full dataset
df['duration_num'] = df['duration'].str.extract(r'(\d+)').astype(float)

# Remove rows with missing duration values
df = df.dropna(subset=['duration_num', 'release_year'])

# 1. Standardise release_year (Mean = 0, Std = 1)
scaler_std = StandardScaler()

df['release_year_standardized'] = scaler_std.fit_transform(
    df[['release_year']]
)

# 2. Normalise duration_num (Scale between 0 and 1)
scaler_minmax = MinMaxScaler()

df['duration_normalized'] = scaler_minmax.fit_transform(
    df[['duration_num']]
)

# Display results
df[['release_year',
    'release_year_standardized',
    'duration_num',
    'duration_normalized']].head()
```

Standardisation (Z-Score Scaling): Applied to release_year to transform it into a distribution with a mean of 0 and a standard deviation of 1. Standardisation works best for bounded, bell-curved, or historical time features.

# Part F: **Sampling**

```python
import pandas as pd

# Load dataset
df = pd.read_csv('netflix_titles.csv')

# Prepare numeric duration column
df['duration_num'] = df['duration'].str.extract(r'(\d+)').astype(float)

# 1. Simple Random Sampling (10% of data)
srs_sample = df.sample(
    frac=0.10,
    random_state=42
)

# 2. Stratified Sampling (10% based on 'type')
stratified_sample = df.groupby('type', group_keys=False).sample(
    frac=0.10,
    random_state=42
)

# Display sample sizes
print("Original dataset size:", len(df))
print("Simple Random Sample size:", len(srs_sample))
print("Stratified Sample size:", len(stratified_sample))

# Check distribution of type in original and stratified samples
print("\nOriginal Type Distribution:")
print(df['type'].value_counts(normalize=True))

print("\nStratified Sample Type Distribution:")
print(stratified_sample['type'].value_counts(normalize=True))
```

```python
# Comparison table for release_year
release_year_stats = pd.DataFrame({
    'Population': [df['release_year'].mean(), df['release_year'].std(), df['release_year'].median()],
    'SRS Sample': [srs_sample['release_year'].mean(), srs_sample['release_year'].std(), srs_sample['release_year'].median()],
    'Stratified Sample': [stratified_sample['release_year'].mean(), stratified_sample['release_year'].std(), stratified_sample['release_year'].median()]
}, index=['Mean', 'Std Dev', 'Median'])

release_year_stats
```

```python
# Comparison table for duration_num
duration_stats = pd.DataFrame({
    'Population': [df['duration_num'].mean(), df['duration_num'].std(), df['duration_num'].median()],
    'SRS Sample': [srs_sample['duration_num'].mean(), srs_sample['duration_num'].std(), srs_sample['duration_num'].median()],
    'Stratified Sample': [stratified_sample['duration_num'].mean(), stratified_sample['duration_num'].std(), stratified_sample['duration_num'].median()]
}, index=['Mean', 'Std Dev', 'Median'])

duration_stats
```

# **Part G: Conclusion**

## 1. Major Insights Obtained
* **Content Type Distribution:** The Netflix catalog is heavily dominated by Movies, which constitute approximately **69.6%** (6,131 titles) of the total library, while TV Shows represent **30.4%** (2,676 titles).
* **Temporal Patterns:** Content additions are strongly skewed toward recent years, showing a sharp increase in catalog additions released between **2015 and 2021** (with a median release year of 2017).
* **Runtime Trends:** The mean duration for movies is **99.5 minutes** (with a mode of 90 minutes), whereas TV shows are predominantly single-season titles.

## 2. Data Quality Issues
* **Missing Value Imputation:** Significant missing data was present in descriptive metadata columns—specifically `director` (**2,634 missing**, ~30%) and `cast` (**825 missing**, ~9.4%). Replacing these missing values with `'Unknown'` prevents unnecessary dropping of dataset rows.
* **Column Misalignment:** Field misalignments were detected (e.g., *Louis C.K.* comedy specials) where duration values like `"74 min"` were erroneously placed into the `rating` column while the `duration` column remained missing.
* **Format Standardization:** The mixed string representations in `duration` (`"90 min"` for movies vs `"2 Seasons"` for TV shows) required regex extraction to convert values into numerical attributes prior to statistical and machine learning applications.

## 3. Important Relationships Between Variables
* **Duration vs. Content Type:** Movie runtimes follow a near-normal distribution centered around ~99 minutes, whereas TV Shows exhibit a right-skewed decay pattern where 1-season shows represent the overwhelming majority.
* **Target Demographics (Ratings):** Ratings like `TV-MA` and `TV-14` make up the largest overall share across both content types, highlighting Netflix's focus on mature and young-adult audiences.
* **Release Year vs. Platform Addition:** A wider gap exists between original release dates and platform addition dates for older movies, whereas newer TV series demonstrate much tighter alignment between release year and Netflix addition year.
